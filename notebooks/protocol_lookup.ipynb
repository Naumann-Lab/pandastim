```python
og = pd.read_hdf(og_path)
```

<h1>Make overlapping stim file<h1>

```python
dot = {'stim_name': ['right_dot'],
          'angle': [90],
          'velocity': [0.02],
          'stim_type': ['s'],
          'stationary_time': [3],
          'duration': [35],
           'texture': ['gray_circle'],
          'center': [[-100, 20]]}
dot_df = pd.DataFrame.from_dict(dot)

lateral_right = {'stim_name': ['lateral_right'],
          'angle': [(270, 90)],
          'velocity': [(0, 0.02)],
          'stim_type': ['b'],
          'stationary_time': [3],
          'duration': [35],
           'texture': ['grating_rgb'],
                 'center': [([0,0], [0,0])]}
lateral_right_df = pd.DataFrame.from_dict(lateral_right)
```

```python
new = pd.DataFrame()
og['texture'] =  ['grating_rgb'] * len(og)
og['center'] =  [[0, 0]] * len(og)
for wholefield_row in og.index:
    wholefield_df = og.loc[wholefield_row]
    mask = {'stim_name': [[dot['stim_name'][0], wholefield_df['stim_name']]],
          'angle': [[dot['angle'][0], wholefield_df['angle']]],
          'velocity': [[dot['velocity'][0], wholefield_df['velocity']]],
          'stim_type': [['m', 'm']],
          'stationary_time': [[dot['stationary_time'][0], wholefield_df['stationary_time']]],
          'duration': [wholefield_df['duration']],
            'texture': [[dot['texture'][0], wholefield_df['texture']]],
            'center': [[dot['center'][0], wholefield_df['center']]]}
    mask_df = pd.DataFrame.from_dict(mask)
    new = pd.concat([new, lateral_right_df,pd.DataFrame(og.loc[wholefield_row]).T, dot_df, mask_df], axis=0).reset_index(drop = True)
    #new = pd.concat([new, mask_df, dot_df], axis=0).reset_index(drop = True)
```
